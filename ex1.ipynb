```python
plt.figure(figsize=(15,10))

sns.heatmap(df.corr(numeric_only=True),
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["Total Waste (Tons)"],
             bins=30,
             kde=True)

plt.title("Distribution of Total Waste (Tons)")
plt.xlabel("Total Waste (Tons)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x="Food Category")

plt.xticks(rotation=45)

plt.title("Food Category Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    y="Country",
    order=df["Country"].value_counts().index[:10]
)

plt.title("Top 10 Countries")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Total Waste (Tons)"])

plt.title("Outliers in Total Waste (Tons)")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Avg Waste per Capita (Kg)",
    y="Total Waste (Tons)"
)

plt.title("Average Waste per Capita vs Total Waste")
plt.xlabel("Average Waste per Capita (Kg)")
plt.ylabel("Total Waste (Tons)")

plt.show()
```

```python
sns.pairplot(df[[
    "Total Waste (Tons)",
    "Economic Loss (Million $)",
    "Avg Waste per Capita (Kg)",
    "Population (Million)",
    "Household Waste (%)"
]])

plt.show()
```

```python
country = df.groupby("Country")["Total Waste (Tons)"].mean().sort_values(ascending=False).head(10)

country.plot(kind="bar")

plt.ylabel("Average Total Waste (Tons)")
plt.title("Top 10 Countries by Average Food Waste")

plt.show()
```

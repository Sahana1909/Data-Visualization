```python
# ============================================================
# ELECTRICITY CONSUMPTION ANALYSIS AND DATA VISUALIZATION
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_style("whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_name = "household_power_consumption.csv"

# Automatically detect whether the file uses comma or semicolon
with open(file_name, "r", encoding="utf-8", errors="ignore") as f:
    first_line = f.readline()

if ";" in first_line:
    separator = ";"
else:
    separator = ","

df = pd.read_csv(
    file_name,
    sep=separator,
    na_values=["?", ""],
    low_memory=False
)

print("\nDataset loaded successfully!")


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\n================ FIRST 5 ROWS ================")

display(df.head())


# ============================================================
# 4. DISPLAY LAST 5 ROWS
# ============================================================

print("\n================ LAST 5 ROWS ================")

display(df.tail())


# ============================================================
# 5. DATASET SIZE
# ============================================================

print("\n================ DATASET SIZE ================")

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])


# ============================================================
# 6. COLUMN NAMES
# ============================================================

print("\n================ COLUMN NAMES ================")

print(df.columns.tolist())


# ============================================================
# 7. DATA INFORMATION
# ============================================================

print("\n================ DATA INFORMATION ================")

df.info()


# ============================================================
# 8. DESCRIPTIVE STATISTICS
# ============================================================

print("\n================ DESCRIPTIVE STATISTICS ================")

display(df.describe())


# ============================================================
# 9. MISSING VALUES
# ============================================================

print("\n================ MISSING VALUES ================")

missing_values = df.isnull().sum()

display(missing_values)


# ============================================================
# 10. MISSING VALUE PERCENTAGE
# ============================================================

print("\n================ MISSING VALUE PERCENTAGE ================")

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

display(missing_percentage)


# ============================================================
# 11. DUPLICATE VALUES
# ============================================================

print("\n================ DUPLICATE ROWS ================")

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# ============================================================
# 12. UNIQUE VALUES
# ============================================================

print("\n================ UNIQUE VALUES ================")

display(df.nunique())


# ============================================================
# 13. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\nCleaned column names:")

print(df.columns.tolist())


# ============================================================
# 14. CONVERT DATE AND TIME
# ============================================================

if "Date" in df.columns and "Time" in df.columns:

    df["DateTime"] = pd.to_datetime(
        df["Date"].astype(str) + " " + df["Time"].astype(str),
        dayfirst=True,
        errors="coerce"
    )

elif "DateTime" in df.columns:

    df["DateTime"] = pd.to_datetime(
        df["DateTime"],
        errors="coerce"
    )

else:

    print("\nDate and Time columns were not found.")


# ============================================================
# 15. CONVERT NUMERIC COLUMNS
# ============================================================

possible_numeric_columns = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

numeric_columns = []

for column in possible_numeric_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        numeric_columns.append(column)


print("\n================ NUMERIC COLUMNS ================")

print(numeric_columns)


# ============================================================
# 16. MISSING VALUES AFTER CONVERSION
# ============================================================

print("\n================ MISSING VALUES AFTER CONVERSION ================")

display(df[numeric_columns].isnull().sum())


# ============================================================
# 17. HANDLE MISSING VALUES
# ============================================================

# Forward fill followed by backward fill
# This is suitable for time-series electricity data.

if len(numeric_columns) > 0:

    df[numeric_columns] = (
        df[numeric_columns]
        .ffill()
        .bfill()
    )


print("\nMissing values handled successfully!")


# ============================================================
# 18. CHECK CLEANED DATA
# ============================================================

print("\n================ CLEANED DATA ================")

display(df.head())

print("\nRemaining missing values:")

display(df[numeric_columns].isnull().sum())


# ============================================================
# 19. CORRELATION MATRIX
# ============================================================

print("\n================ CORRELATION MATRIX ================")

correlation = df[numeric_columns].corr()

display(correlation)


# ============================================================
# 20. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Correlation Between Electricity Consumption Variables"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. GLOBAL ACTIVE POWER DISTRIBUTION
# ============================================================

if "Global_active_power" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df["Global_active_power"],
        bins=40,
        kde=True
    )

    plt.title(
        "Distribution of Global Active Power"
    )

    plt.xlabel(
        "Global Active Power"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 22. GLOBAL REACTIVE POWER DISTRIBUTION
# ============================================================

if "Global_reactive_power" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df["Global_reactive_power"],
        bins=40,
        kde=True
    )

    plt.title(
        "Distribution of Global Reactive Power"
    )

    plt.xlabel(
        "Global Reactive Power"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 23. VOLTAGE DISTRIBUTION
# ============================================================

if "Voltage" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df["Voltage"],
        bins=40,
        kde=True
    )

    plt.title(
        "Distribution of Voltage"
    )

    plt.xlabel(
        "Voltage (V)"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 24. GLOBAL INTENSITY DISTRIBUTION
# ============================================================

if "Global_intensity" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df["Global_intensity"],
        bins=40,
        kde=True
    )

    plt.title(
        "Distribution of Global Intensity"
    )

    plt.xlabel(
        "Global Intensity (A)"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 25. BOXPLOT - ACTIVE POWER
# ============================================================

if "Global_active_power" in df.columns:

    plt.figure(figsize=(10, 5))

    sns.boxplot(
        x=df["Global_active_power"]
    )

    plt.title(
        "Global Active Power - Outlier Analysis"
    )

    plt.xlabel(
        "Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 26. BOXPLOT - VOLTAGE
# ============================================================

if "Voltage" in df.columns:

    plt.figure(figsize=(10, 5))

    sns.boxplot(
        x=df["Voltage"]
    )

    plt.title(
        "Voltage - Outlier Analysis"
    )

    plt.xlabel(
        "Voltage (V)"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 27. BOXPLOT - GLOBAL INTENSITY
# ============================================================

if "Global_intensity" in df.columns:

    plt.figure(figsize=(10, 5))

    sns.boxplot(
        x=df["Global_intensity"]
    )

    plt.title(
        "Global Intensity - Outlier Analysis"
    )

    plt.xlabel(
        "Global Intensity (A)"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 28. ACTIVE POWER VS GLOBAL INTENSITY
# ============================================================

if (
    "Global_active_power" in df.columns
    and "Global_intensity" in df.columns
):

    sample_df = df.sample(
        min(5000, len(df)),
        random_state=42
    )

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=sample_df,
        x="Global_intensity",
        y="Global_active_power"
    )

    plt.title(
        "Global Intensity vs Global Active Power"
    )

    plt.xlabel(
        "Global Intensity (A)"
    )

    plt.ylabel(
        "Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 29. ACTIVE POWER VS VOLTAGE
# ============================================================

if (
    "Global_active_power" in df.columns
    and "Voltage" in df.columns
):

    sample_df = df.sample(
        min(5000, len(df)),
        random_state=42
    )

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=sample_df,
        x="Voltage",
        y="Global_active_power"
    )

    plt.title(
        "Voltage vs Global Active Power"
    )

    plt.xlabel(
        "Voltage (V)"
    )

    plt.ylabel(
        "Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 30. REACTIVE POWER VS ACTIVE POWER
# ============================================================

if (
    "Global_active_power" in df.columns
    and "Global_reactive_power" in df.columns
):

    sample_df = df.sample(
        min(5000, len(df)),
        random_state=42
    )

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=sample_df,
        x="Global_reactive_power",
        y="Global_active_power"
    )

    plt.title(
        "Reactive Power vs Active Power"
    )

    plt.xlabel(
        "Global Reactive Power"
    )

    plt.ylabel(
        "Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 31. CREATE TIME FEATURES
# ============================================================

if "DateTime" in df.columns:

    df["Year"] = df["DateTime"].dt.year

    df["Month"] = df["DateTime"].dt.month

    df["Month_Name"] = df["DateTime"].dt.month_name()

    df["Day"] = df["DateTime"].dt.day

    df["Day_Name"] = df["DateTime"].dt.day_name()

    df["Hour"] = df["DateTime"].dt.hour


# ============================================================
# 32. DAILY ELECTRICITY CONSUMPTION
# ============================================================

if (
    "DateTime" in df.columns
    and "Global_active_power" in df.columns
):

    daily_power = (
        df.set_index("DateTime")
        ["Global_active_power"]
        .resample("D")
        .mean()
    )

    plt.figure(figsize=(14, 6))

    daily_power.plot()

    plt.title(
        "Daily Average Electricity Consumption"
    )

    plt.xlabel(
        "Date"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 33. MONTHLY ELECTRICITY CONSUMPTION
# ============================================================

if (
    "DateTime" in df.columns
    and "Global_active_power" in df.columns
):

    monthly_power = (
        df.set_index("DateTime")
        ["Global_active_power"]
        .resample("ME")
        .mean()
    )

    plt.figure(figsize=(14, 6))

    monthly_power.plot(
        marker="o"
    )

    plt.title(
        "Monthly Average Electricity Consumption"
    )

    plt.xlabel(
        "Month"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 34. HOURLY ELECTRICITY CONSUMPTION
# ============================================================

if (
    "DateTime" in df.columns
    and "Global_active_power" in df.columns
):

    hourly_power = (
        df.set_index("DateTime")
        ["Global_active_power"]
        .resample("h")
        .mean()
    )

    plt.figure(figsize=(14, 6))

    hourly_power.plot()

    plt.title(
        "Hourly Electricity Consumption Pattern"
    )

    plt.xlabel(
        "Time"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 35. AVERAGE CONSUMPTION BY DAY OF WEEK
# ============================================================

if (
    "Day_Name" in df.columns
    and "Global_active_power" in df.columns
):

    day_order = [
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday"
    ]

    day_power = (
        df.groupby("Day_Name")
        ["Global_active_power"]
        .mean()
        .reindex(day_order)
    )

    print(
        "\n================ AVERAGE POWER BY DAY ================"
    )

    display(day_power)


# ============================================================
# 36. DAY-WISE BAR CHART
# ============================================================

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=day_power.index,
        y=day_power.values
    )

    plt.title(
        "Average Electricity Consumption by Day"
    )

    plt.xlabel(
        "Day of Week"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 37. MONTH-WISE CONSUMPTION
# ============================================================

if (
    "Month_Name" in df.columns
    and "Global_active_power" in df.columns
):

    month_order = [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July",
        "August",
        "September",
        "October",
        "November",
        "December"
    ]

    month_power = (
        df.groupby("Month_Name")
        ["Global_active_power"]
        .mean()
        .reindex(month_order)
    )

    print(
        "\n================ AVERAGE POWER BY MONTH ================"
    )

    display(month_power)


# ============================================================
# 38. MONTH-WISE BAR CHART
# ============================================================

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=month_power.index,
        y=month_power.values
    )

    plt.title(
        "Average Electricity Consumption by Month"
    )

    plt.xlabel(
        "Month"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 39. HOURLY PATTERN
# ============================================================

if (
    "Hour" in df.columns
    and "Global_active_power" in df.columns
):

    hour_power = (
        df.groupby("Hour")
        ["Global_active_power"]
        .mean()
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        x=hour_power.index,
        y=hour_power.values,
        marker="o"
    )

    plt.title(
        "Average Electricity Consumption by Hour"
    )

    plt.xlabel(
        "Hour of Day"
    )

    plt.ylabel(
        "Average Global Active Power"
    )

    plt.xticks(range(0, 24))

    plt.tight_layout()

    plt.show()


# ============================================================
# 40. SUB-METERING ANALYSIS
# ============================================================

sub_meter_columns = [
    column
    for column in [
        "Sub_metering_1",
        "Sub_metering_2",
        "Sub_metering_3"
    ]
    if column in df.columns
]

if len(sub_meter_columns) > 0:

    sub_metering_average = (
        df[sub_meter_columns]
        .mean()
    )

    print(
        "\n================ AVERAGE SUB-METERING ================"
    )

    display(sub_metering_average)


# ============================================================
# 41. SUB-METERING BAR CHART
# ============================================================

if len(sub_meter_columns) > 0:

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=sub_metering_average.index,
        y=sub_metering_average.values
    )

    plt.title(
        "Average Electricity Usage by Sub-Meter"
    )

    plt.xlabel(
        "Sub-Meter"
    )

    plt.ylabel(
        "Average Consumption"
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 42. TOP 10 HIGHEST POWER RECORDS
# ============================================================

if "Global_active_power" in df.columns:

    top_power = (
        df.sort_values(
            by="Global_active_power",
            ascending=False
        )
        .head(10)
    )

    print(
        "\n================ TOP 10 HIGHEST POWER RECORDS ================"
    )

    display(
        top_power[
            [
                column
                for column in [
                    "Date",
                    "Time",
                    "Global_active_power",
                    "Voltage",
                    "Global_intensity"
                ]
                if column in top_power.columns
            ]
        ]
    )


# ============================================================
# 43. TOP 10 HIGHEST POWER VISUALIZATION
# ============================================================

if (
    "Global_active_power" in df.columns
    and "Date" in df.columns
):

    plt.figure(figsize=(12, 6))

    sns.barplot(
        data=top_power,
        x="Date",
        y="Global_active_power"
    )

    plt.title(
        "Top 10 Highest Electricity Consumption Records"
    )

    plt.xlabel(
        "Date"
    )

    plt.ylabel(
        "Global Active Power"
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 44. CONSUMPTION BY DAY - BOXPLOT
# ============================================================

if (
    "Day_Name" in df.columns
    and "Global_active_power" in df.columns
):

    plt.figure(figsize=(12, 6))

    sns.boxplot(
        data=df,
        x="Day_Name",
        y="Global_active_power",
        order=day_order
    )

    plt.title(
        "Electricity Consumption Distribution by Day"
    )

    plt.xlabel(
        "Day"
    )

    plt.ylabel(
        "Global Active Power"
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 45. PAIRPLOT
# ============================================================

pair_columns = [
    column
    for column in [
        "Global_active_power",
        "Global_reactive_power",
        "Voltage",
        "Global_intensity"
    ]
    if column in df.columns
]

if len(pair_columns) >= 2:

    pairplot_data = (
        df[pair_columns]
        .dropna()
        .sample(
            min(2000, len(df)),
            random_state=42
        )
    )

    sns.pairplot(
        pairplot_data
    )

    plt.show()


# ============================================================
# 46. FINAL STATISTICAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)

print(
    "       ELECTRICITY CONSUMPTION ANALYSIS SUMMARY"
)

print("=" * 60)


print(
    "\nTotal Records :",
    len(df)
)

print(
    "Total Columns :",
    len(df.columns)
)


if "Global_active_power" in df.columns:

    print(
        "\nAverage Active Power :",
        round(
            df["Global_active_power"].mean(),
            2
        )
    )

    print(
        "Minimum Active Power :",
        round(
            df["Global_active_power"].min(),
            2
        )
    )

    print(
        "Maximum Active Power :",
        round(
            df["Global_active_power"].max(),
            2
        )
    )


if "Global_reactive_power" in df.columns:

    print(
        "Average Reactive Power :",
        round(
            df["Global_reactive_power"].mean(),
            2
        )
    )


if "Voltage" in df.columns:

    print(
        "Average Voltage :",
        round(
            df["Voltage"].mean(),
            2
        )
    )


if "Global_intensity" in df.columns:

    print(
        "Average Global Intensity :",
        round(
            df["Global_intensity"].mean(),
            2
        )
    )


print(
    "\nDuplicate Rows :",
    df.duplicated().sum()
)


print(
    "\nRemaining Missing Values :",
    df.isnull().sum().sum()
)


print("\n")
print("=" * 60)

print(
    "EDA AND DATA VISUALIZATION COMPLETED SUCCESSFULLY!"
)

print("=" * 60)
```
